```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
data1 = np.linspace(0,10,20)
```

```python
data1
```

```python
data2 = data1 ** 2 
```

```python
data2
```

```python
plt.subplots()
```

```python
my_fig , my_axes = plt.subplots()
```

```python
type(my_fig)
```

```python
type(my_axes)
```

```python
(my_fig , my_axes) = plt.subplots()
my_axes.plot(data1 , data2)
my_axes.plot(data2 , data1)
plt.show()
```

```python
(new_fig , new_axes) = plt.subplots()

new_axes.plot(data1 , data1 + 2 , color="blue", linewidth=1.0)
new_axes.plot(data1 , data1 + 3 , color="red", linewidth=3.0)
new_axes.plot(data1 , data1 + 4 , color="green", linestyle= "-.")

plt.show()
```
