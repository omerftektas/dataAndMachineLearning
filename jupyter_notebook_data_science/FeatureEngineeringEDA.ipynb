```python
df__app_category = df_clean.groupby(["Category","App"])["Installs"].sum().reset_index()
```

```python
df__app_category
```

```python
df_app_category = df__app_category.sort_values("Installs", ascending=False)
```

```python
apps = ["GAME","COMMUNICATION","TOOLS","PRODUCTIVITY","SOCIAL"]
plt.figure(figsize=(40,30))
for i, app in enumerate(apps):
    df2 = df_app_category[df_app_category["Category"] == app]
    df2 = df2.head(5)
    #print(df2)

    plt.subplot(5,2, i+1)
    sns.barplot(data = df2, x="Installs", y="App")
    plt.title(app, size=20)
plt.tight_layout()
plt.show()
```
