```python
def salary_category(salary):
    if salary < 50000:
        return "low"
    elif 50000 <= salary < 80000:
        return "medium"
    else:
        return "high"
```

```python
df["Salary"].apply(salary_category)
```

```python
df["Salary_Category"] = df["Salary"].apply(salary_category)
```

```python
df
```

```python
def new_perf(experience):
    if experience > 10:
        return 1
    else:
        return 0
```

```python
df["new_perf"] = df["Experience"].apply(new_perf)
```

```python
df
```

```python
df["New Last Score"] = df["Performance_Score"] + df["new_perf"]
```

```python
df
```

```python
def adjust_new(row):
    if row["Experience"] > 10:
        return row["Performance_Score"] + 1 
    else:
        return row["Performance_Score"]
```

```python
df["Adjusted Score"] = df.apply(adjust_new, axis = 1)
```

```python
df
```

```python
df["Formated Name"] = df["Name"].apply(lambda x : x.replace("_", " "))
```
