```python
plt.figure(figsize=(10,9))
sns.heatmap(data.corr())
```

```python
sns.set_theme
sns.scatterplot(x="alcohol", y="quality", data = data)
plt.xlabel = "Alcohol unit "
plt.ylabel= "Quality 1-10"
plt.title = "Alcohol and Quality Relation"
plt.show()
```

```python
sns.set_theme
sns.scatterplot(x="alcohol", y="quality", hue = "pH", data = data)
plt.xlabel = "Alcohol unit "
plt.ylabel= "quality unit"
plt.title = "Alcohol and Quality Relation"
plt.show()
```

```python
sns.set_theme
sns.lineplot(x="alcohol", y="quality", data = data)
plt.xlabel = "Alcohol unit "
plt.ylabel= "Quality 1-10"
plt.title = "Alcohol and Quality Relation"
plt.show()
```

```python
sns.set_theme
sns.displot(x="volatile acidity", y="quality",data=data)
plt.show()
```

```python
sns.set_theme
sns.lineplot(x="volatile acidity", y="quality",data=data)
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.boxplot(y=data["quality"])
plt.show()
```

```python
plt.subplot(1,2,1)
sns.boxplot(y="pH",data=data)
plt.subplot(1,2,2)
sns.boxplot(y="citric acid",data=data)


plt.tight_layout()#daha duzgun layout olsun dıye
plt.show()
```

```python
sns.set_theme
sns.lineplot(x="fixed acidity",y="citric acid",data=data)
plt.show()
```

```python
sns.set_theme
sns.boxplot(y="alcohol",data = data)
plt.show()
```

```python
#plt.figure(figsize=(8, 4))
sns.countplot(x='quality', data=data)
plt.title('Şarap Kalite Puanlarının Dağılımı')
plt.xlabel('Kalite Puanı')
plt.ylabel('Örnek Sayısı')
plt.show()
```

```python
columns = data.columns
(fig,ax)= plt.subplots(4,4,figsize=(16,15))## 4 row 4 kolon
ax = ax.flatten() ## dızımızı tek boyuta ındırgedık

for i, column in enumerate(columns):
    sns.kdeplot(
       data=data,
        x = column,
        hue = data.quality,
        ax = ax[i]
    )
    ax[i].set_title(f"{column } Distribution")
    ax[i].set_xlabel(None)
    
for i in range(i+1, len(ax)): ## bos ax ları silme kodu
    ax[i].axis("off")
```
